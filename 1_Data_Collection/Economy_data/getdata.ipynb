Yahoo Finance is a popular online platform that provides financial news, data, and analysis. It offers tools for tracking stock prices, market trends, and economic indicators. Users can access historical and real-time data for stocks, indices, currencies, and other financial instruments. Yahoo Finance also provides APIs, such as `yfinance`, which allow developers to programmatically download and analyze financial data for research and decision-making.

````python
import yfinance as yf
import pandas as pd
import datetime
````

````python
# Download IHSG (^JKSE) daily data
end_date = datetime.date.today().isoformat()
ihsg = yf.download("^JKSE", start="1995-01-01", end=end_date)
````

````python
# Define the ticker symbol for USD/IDR
ticker_symbol = "IDR=X"

# Create a ticker object
usd_idr = yf.Ticker(ticker_symbol)

# Get historical data for the maximum possible period
hist_data = usd_idr.history(period="max")

# Check if data was successfully fetched
if hist_data.empty:
    print(f"No data found for ticker {ticker_symbol}. Please check the ticker symbol.")
else:
    print(f"Successfully fetched data from {hist_data.index.min().date()} to {hist_data.index.max().date()}.")

hist_data.to_csv("usd_idr_daily.csv")
````

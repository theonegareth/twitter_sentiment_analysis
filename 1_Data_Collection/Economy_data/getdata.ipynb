```python
import yfinance as yf
import pandas as pd
```

```python
# Download IHSG (^JKSE) daily data
ihsg = yf.download("^JKSE", start="1995-01-01", end="2025-10-07")

# Display first rows
print(ihsg.head())

# Optionally save to CSV
ihsg.to_csv("ihsg_daily.csv")
```

```python
# Define the ticker symbol for USD/IDR
ticker_symbol = "IDR=X"

# Create a ticker object
usd_idr = yf.Ticker(ticker_symbol)

# Get historical data for the maximum possible period
hist_data = usd_idr.history(period="max")

# Check if data was successfully fetched
if hist_data.empty:
    print(f"No data found for ticker {ticker_symbol}. Please check the ticker symbol.")
else:
    print(f"Successfully fetched data from {hist_data.index.min().date()} to {hist_data.index.max().date()}.")

hist_data.to_csv("usd_idr_daily.csv")
```
